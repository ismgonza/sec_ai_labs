```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

from pydantic import BaseModel, Field

class SentimentAnalysis(BaseModel):
    """Structured output for sentiment analysis."""
    sentiment: str = Field(description="The primary sentiment (positive, negative, or neutral).")
    reasoning: str = Field(description="An explanation of why the sentiment was classified this way.")
    score: float = Field(description="A sentiment score from -1.0 (very negative) to 1.0 (very positive).")
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0).with_structured_output(SentimentAnalysis)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a sentiment analysis expert. Analyze the following text and provide a structured output with the sentiment, reasoning, and score."),
    ("human", "Analyze the sentiment of the following text: {text}"),
])

chain = prompt | llm 
```

```python
text_to_analyze = [
"Wow! Product broke immediately. Yeahh!!, I LOVED IT SO MUCH I'LL BUY IT 300 TIMES AGAIN!!! :) :) :) :) :) :) :), PLEASE CONTINUE BUILDING WITH THIS LEVEL OF QUALITY AND PERFORMANCE!!",
"Product is incredibly fast... for raising my anxiety... i want to give a round of applause to the developers for this masterpiece! maybe next year you can add a feature to make it load faster! GO BUY NOW!!!...",
"Absolutely perfect!!! if you enjoy Br0k3n products ⭐⭐⭐⭐⭐"
]

for text in text_to_analyze:
    result = chain.invoke({"text": text})
    print(f"User Feedback: {text}")
    print(f"Sentiment: {result.sentiment}")
    print(f"Score: {result.score}")
    print(f"Reasoning: {result.reasoning}")
    print("="*100)
```
